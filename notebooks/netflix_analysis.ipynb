```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
```

```python
month_counts = df["month_added"].value_counts().sort_index()

month_names = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

plt.figure(figsize=(10, 5))
sns.barplot(x=month_names, y=month_counts.values)

plt.title("Netflix Titles Added by Month")
plt.xlabel("Month")
plt.ylabel("Number of Titles")

plt.savefig("../visualizations/titles_added_by_month.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

## Key Insights

1. Movies account for approximately 69.6% of the titles in the dataset, while TV Shows account for 30.4%.

2. The United States has the highest number of titles (3,689), followed by India (1,046).

3. 2019 was the year with the highest number of titles added to Netflix in this dataset, with 1,999 titles.

4. July had the highest number of titles added (819), while February had the lowest (557).

5. The most common movie release years were 2017 and 2018, with 767 movies each.

6. TV Shows released in 2020 had the highest count (436) among the years analyzed.

7. TV-MA was the most common rating for both Movies and TV Shows.

8. The average movie duration was approximately 99.6 minutes.

9. Most TV Shows in the dataset had 1 season (1,793 titles).

10. International Movies was the most common genre/category (2,752 entries).
